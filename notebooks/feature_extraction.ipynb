```python
from master_thesis.feature_extraction import extract_features_batch, standardize_features
```

```python
from master_thesis.constants import COLUMNS

train_waveforms = "../data/EchoNext_train_waveforms.npy"
val_waveforms = "../data/EchoNext_val_waveforms.npy"
test_waveforms = "../data/EchoNext_test_waveforms.npy"

train_tabular = "../data/EchoNext_train_tabular_features.npy"
val_tabular = "../data/EchoNext_val_tabular_features.npy"
test_tabular = "../data/EchoNext_test_tabular_features.npy"

metadata_path = "../data/echonext_metadata_100k.csv"

# Columns to skip when scaling (already pre-scaled tabular features)
TABULAR_COLS = COLUMNS
```

```python
train_features = extract_features_batch(train_waveforms, tabular_features_path=train_tabular, metadata_path=metadata_path, split="train")
train_features_standardized = standardize_features(train_features, fit=True, scaler_path="../processed_data/feature_scaler.joblib", exclude_cols=TABULAR_COLS)
train_features.to_csv("../processed_data/train_features.csv", index=False)
train_features_standardized.to_csv("../processed_data/train_features_standardized.csv", index=False)
```

```python
ecg_feature_cols = [
    "rr_mean", "rr_std", "rmssd", 
    "r_amp_mean", "r_amp_std", 
    "qrs_energy_mean", "qrs_energy_std", 
    "signal_std", "signal_skew", "signal_kurtosis", 
    "fft_energy"
]

num_dropped = train_features[ecg_feature_cols].isna().all(axis=1).sum()
num_total = len(train_features)
num_valid = num_total - num_dropped

print("=" * 40)
print("       DATA DROPPED / CORRUPTION REPORT")
print("=" * 40)
print(f"Total Waveforms in Array  : {num_total:,}")
print(f"Clean & Valid Waveforms   : {num_valid:,} ({(num_valid/num_total)*100:.2f}%)")
print(f"Dropped Waveforms (NaNs)  : {num_dropped:,} ({(num_dropped/num_total)*100:.2f}%)")
print("=" * 40)
```

```python
val_features = extract_features_batch(val_waveforms, tabular_features_path=val_tabular, metadata_path=metadata_path, split="val")
val_features_standardized = standardize_features(val_features, fit=False, scaler_path="../processed_data/feature_scaler.joblib", exclude_cols=TABULAR_COLS)
val_features.to_csv("../processed_data/val_features.csv", index=False)
val_features_standardized.to_csv("../processed_data/val_features_standardized.csv", index=False)
```
