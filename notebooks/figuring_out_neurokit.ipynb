```python
import neurokit2 as nk
import numpy as np
import pandas as pd
```

```python
example_waveform = np.load("../data/EchoNext_train_waveforms.npy", mmap_mode="r")[2]
example_waveform.shape
```

```python
one_lead = example_waveform[0, :, 1] # second lead r peaks are the most visible
one_lead.shape
```

```python
signals, info = nk.ecg_process(one_lead, sampling_rate=250)
```

```python
nk.ecg_plot(signals, info)
```

```python
info
```

```python
analyze_df = nk.ecg_analyze(signals, sampling_rate=250, method="interval-related")
```

```python
analyze_df.columns
```

```python
analyze_df
```

```python
# both are due to the fact that the signal is too short (only 10 seconds, which is not enough for some features to be calculated)
# it can be written in thesis as limitation that in the future someone might want to use longer signals to get more features calculated
# remove nan
analyze_df_non_nan = analyze_df.loc[:, ~analyze_df.apply(lambda c: c.map(lambda x: isinstance(x, (list, tuple, np.ndarray)) and np.asarray(x, dtype=float).size > 0 and np.isnan(np.asarray(x, dtype=float)).all()).all())].copy()
# remove inf or -inf
analyze_df_non_nan = analyze_df_non_nan.loc[:, ~analyze_df_non_nan.apply(lambda c: c.map(lambda x: isinstance(x, (list, tuple, np.ndarray)) and np.asarray(x, dtype=float).size > 0 and np.isinf(np.asarray(x, dtype=float)).all()).all())].copy()
```

```python
analyze_df_non_nan
```

```python
for col in analyze_df_non_nan.columns:
    analyze_df_non_nan[col] = analyze_df_non_nan[col].apply(lambda x: np.squeeze(x) if isinstance(x, (list, np.ndarray)) else x)
```
